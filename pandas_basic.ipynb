## 1. Pandas Series

A Series is a one-dimensional labeled data structure.

```python
sales = pd.Series([100, 200, 150, 300, 250])

sales
```

```python
print("First value:", sales[0])
print("Fourth value:", sales[3])
```

## 2. Creating a DataFrame

A DataFrame is a two-dimensional table consisting of rows and columns.

```python
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Price": [55000, 800, 1500, 12000],
    "Quantity": [2, 5, 3, 2]
}

df = pd.DataFrame(data)

df
```

## 3. Inspecting the DataFrame

```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Shape:", df.shape)

print("Columns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
df.info()
```

```python
df.describe()
```

## 4. Selecting Columns

```python
# Single column
df["Product"]
```

```python
# Multiple columns
df[["Product", "Price"]]
```

## 5. Selecting Rows

### iloc
`iloc` selects rows and columns using integer positions.

### loc
`loc` selects data using labels or conditions.

```python
# First row
df.iloc[0]
```

```python
# First two rows
df.iloc[0:2]
```

```python
# First row, second column
df.iloc[0, 1]
```

```python
# Select first row using label
df.loc[0]
```

```python
# Products with price greater than 5000
df.loc[df["Price"] > 5000]
```

## 6. Filtering Data

Filtering is one of the most important Pandas skills for Data Analysts.

```python
# Quantity greater than 2
df[df["Quantity"] > 2]
```

```python
# Price greater than 1000 AND quantity greater than 2
df[(df["Price"] > 1000) & (df["Quantity"] > 2)]
```

```python
# Price greater than 50000 OR quantity greater than 4
df[(df["Price"] > 50000) | (df["Quantity"] > 4)]
```

### Logical Operators

- `&` → AND
- `|` → OR
- `~` → NOT

```python
## 7. Sorting Data
```

```python
# Sort by price - ascending
df.sort_values("Price")
```

```python
# Sort by price - descending
df.sort_values("Price", ascending=False)
```

## 8. Creating a New Column

We can create a Sales column using:

Sales = Price × Quantity

```python
df["Sales"] = df["Price"] * df["Quantity"]

df
```

```python
## 9. Basic Statistics
```

```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
```
