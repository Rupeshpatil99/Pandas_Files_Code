```python
import pandas as pd

# Create DataFrame
employee_data = {
    "Name": ["Amit", "Rohit", "Sneha", "Priya", "Karan"],
    "Age": [22, 25, 21, 28, 24],
    "City": ["Pune", "Mumbai", "Nashik", "Pune", "Mumbai"],
    "Salary": [35000, 45000, 30000, 55000, 40000]
}

employees = pd.DataFrame(employee_data)

# Q1. Display first 3 rows
print("Q1. First 3 rows:")
print(employees.head(3))

# Q2. Display only Name and Salary
print("\nQ2. Name and Salary:")
print(employees[["Name", "Salary"]])

# Q3. Salary greater than 40000
print("\nQ3. Salary greater than 40000:")
print(employees[employees["Salary"] > 40000])

# Q4. Employees from Pune
print("\nQ4. Employees from Pune:")
print(employees[employees["City"] == "Pune"])

# Q5. Sort salary in descending order
print("\nQ5. Salary sorted descending:")
print(employees.sort_values(by="Salary", ascending=False))

# Q6. Create AnnualSalary
employees["AnnualSalary"] = employees["Salary"] * 12

print("\nQ6. Annual Salary:")
print(employees)

# Q7. Average salary
print("\nQ7. Average Salary:")
print(employees["Salary"].mean())

# Q8. Maximum salary
print("\nQ8. Maximum Salary:")
print(employees["Salary"].max())

# Q9. Minimum salary
print("\nQ9. Minimum Salary:")
print(employees["Salary"].min())

# Q10. Employees whose age is greater than 23
print("\nQ10. Age greater than 23:")
print(employees[employees["Age"] > 23])
```
